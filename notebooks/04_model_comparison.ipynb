# Notebook 04: Model Performance Benchmarking & Comparison
This notebook visualizes and compares the performance metrics (MAE, RMSE, R², Training Time, Prediction Time) across all four distributed machine learning algorithms: Linear Regression, Random Forest, XGBoost, and LightGBM.

```python
import os
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append('../src')
from utils import get_project_root

# Set modern visual styling
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 12

root_dir = get_project_root()
plots_dir = root_dir / "results" / "plots"
plots_dir.mkdir(parents=True, exist_ok=True)

# Cell 1: Load results/model_comparison.csv
csv_path = root_dir / "results" / "model_comparison.csv"
df_results = pd.read_csv(csv_path)
print(f"Loaded results from: {csv_path}")
```

```python
# Cell 2: Display Table
display(df_results.style.format({
    'training_time_sec': '{:.2f}s',
    'prediction_time_sec': '{:.2f}s',
    'mae': '£{:,.2f}',
    'rmse': '£{:,.2f}',
    'r2': '{:.4f}'
}))
```

```python
# Cell 3: Generate Bar Chart: MAE by Model
plt.figure(figsize=(9, 5))
ax = sns.barplot(x="model", y="mae", data=df_results, palette="Blues_d")
plt.title("Mean Absolute Error (MAE) by Model — Lower is Better", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Model Architecture", fontsize=12)
plt.ylabel("MAE (£)", fontsize=12)
for p in ax.patches:
    ax.annotate(f"£{p.get_height():,.0f}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=11, fontweight='bold', xytext=(0, 5), textcoords='offset points')
plt.tight_layout()
save_path = plots_dir / "mae_comparison.png"
plt.savefig(save_path, dpi=300)
plt.show()
print(f"Saved: {save_path}")
```

```python
# Cell 4: Generate Bar Chart: RMSE by Model
plt.figure(figsize=(9, 5))
ax = sns.barplot(x="model", y="rmse", data=df_results, palette="Reds_d")
plt.title("Root Mean Squared Error (RMSE) by Model — Lower is Better", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Model Architecture", fontsize=12)
plt.ylabel("RMSE (£)", fontsize=12)
for p in ax.patches:
    ax.annotate(f"£{p.get_height():,.0f}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=11, fontweight='bold', xytext=(0, 5), textcoords='offset points')
plt.tight_layout()
save_path = plots_dir / "rmse_comparison.png"
plt.savefig(save_path, dpi=300)
plt.show()
print(f"Saved: {save_path}")
```

```python
# Cell 5: Generate Bar Chart: R² by Model
plt.figure(figsize=(9, 5))
ax = sns.barplot(x="model", y="r2", data=df_results, palette="Greens_d")
plt.title("Coefficient of Determination (R²) by Model — Higher is Better", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Model Architecture", fontsize=12)
plt.ylabel("R² Score", fontsize=12)
for p in ax.patches:
    ax.annotate(f"{p.get_height():.4f}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=11, fontweight='bold', xytext=(0, 5), textcoords='offset points')
plt.tight_layout()
save_path = plots_dir / "r2_comparison.png"
plt.savefig(save_path, dpi=300)
plt.show()
print(f"Saved: {save_path}")
```

```python
# Cell 6: Generate Bar Chart: Training Time by Model
plt.figure(figsize=(9, 5))
ax = sns.barplot(x="model", y="training_time_sec", data=df_results, palette="Purples_d")
plt.title("Training Execution Time (Seconds) by Model", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Model Architecture", fontsize=12)
plt.ylabel("Training Time (s)", fontsize=12)
for p in ax.patches:
    ax.annotate(f"{p.get_height():.2f}s", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=11, fontweight='bold', xytext=(0, 5), textcoords='offset points')
plt.tight_layout()
save_path = plots_dir / "training_time_comparison.png"
plt.savefig(save_path, dpi=300)
plt.show()
print(f"Saved: {save_path}")
```

```python
# Cell 7: Generate Bar Chart: Prediction Time by Model
plt.figure(figsize=(9, 5))
ax = sns.barplot(x="model", y="prediction_time_sec", data=df_results, palette="Oranges_d")
plt.title("Inference / Prediction Time (Seconds) on Test Partition", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Model Architecture", fontsize=12)
plt.ylabel("Prediction Time (s)", fontsize=12)
for p in ax.patches:
    ax.annotate(f"{p.get_height():.2f}s", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=11, fontweight='bold', xytext=(0, 5), textcoords='offset points')
plt.tight_layout()
save_path = plots_dir / "prediction_time_comparison.png"
plt.savefig(save_path, dpi=300)
plt.show()
print(f"Saved: {save_path}")
```

```python
# Cell 8: Print Summary: Best Model by RMSE, Best Model by R², Fastest Training
best_rmse_row = df_results.loc[df_results["rmse"].idxmin()]
best_r2_row = df_results.loc[df_results["r2"].idxmax()]
fastest_train_row = df_results.loc[df_results["training_time_sec"].idxmin()]
fastest_pred_row = df_results.loc[df_results["prediction_time_sec"].idxmin()]

print("=" * 60)
print("🏆 BENCHMARK WINNERS SUMMARY")
print("=" * 60)
print(f"• Lowest RMSE (Top Accuracy):  {best_rmse_row['model']} (£{best_rmse_row['rmse']:,.2f})")
print(f"• Highest R² (Variance Fit):   {best_r2_row['model']} ({best_r2_row['r2']:.4f})")
print(f"• Fastest Training Duration:   {fastest_train_row['model']} ({fastest_train_row['training_time_sec']:.2f}s)")
print(f"• Lowest Inference Latency:    {fastest_pred_row['model']} ({fastest_pred_row['prediction_time_sec']:.2f}s)")
print("=" * 60)
```

### Cell 9: Summary Insights & Strategic Takeaways
1. **Tree-Based vs Linear Baselines**: Gradient Boosted Trees (XGBoost / LightGBM) and Random Forest capture non-linear geographical interactions (district, county, town indices) far more effectively than basic Linear Regression.
2. **Computational Scalability**: Spark's distributed architecture allows parallelizing tree building across worker nodes efficiently while managing memory footprints.
3. **Inference Latency vs Precision Trade-off**: Linear Regression offers instantaneous inference, whereas Gradient Boosting yields substantially tighter confidence bounds and lower error rates for production valuation models.
